```python
import pandas as pd
from fuzzywuzzy import fuzz, process

# Exemple : chargement manuel (à adapter si fichiers CSV ou Excel)
table_mails = pd.DataFrame({
    'Nom': ['abderrazzak', 'Azouggarh', 'benzouina', 'abderrazzak', 'ECHOUIKHI', 'abderrazzak'],
    'Prenom': ['chaimae', 'oussama', 'douae', 'hasnaa', 'Driss', 'Zouhair'],
    'Mail1': [
        'abderrazzak.chaimae@gmail.com',
        'Azouggarh.oussama@gmail.com',
        'abdelaziz.douae@gmail.com',
        'abderrazzak.hasnaa@gmail.com',
        'ECHOUIKHI.Driss@gmail.com',
        'abderrazzak.Zouhair@gmail.com'
    ]
})

table_users = pd.DataFrame({
    'Username': ['DRISS ECHOUIKHI', 'Chaimae ABDERRAZZAK', 'abderrazzak hasnae', 'Benzouina douae'],
    'Login': ['DRISS', 'Chaimae', 'C_hasnae', 'douae'],
    'Mail': [None, None, None, None]  # à remplir
})

# Nettoyage minuscule
def normalize(text):
    return str(text).strip().lower()

# Ajout d'une colonne pour combinaison dans Users
table_users['Fullname'] = table_users['Username'].apply(normalize)

# Parcours des lignes de table_mails
for idx, row in table_mails.iterrows():
    nom = normalize(row['Nom'])
    prenom = normalize(row['Prenom'])

    combinaison1 = f"{prenom} {nom}"
    combinaison2 = f"{nom} {prenom}"

    best_match = None
    best_score = 0
    best_index = -1

    for i, user_row in table_users.iterrows():
        username = user_row['Fullname']

        score1 = fuzz.ratio(combinaison1, username)
        score2 = fuzz.ratio(combinaison2, username)
        score = max(score1, score2)

        if score > best_score:
            best_score = score
            best_match = user_row['Username']
            best_index = i

    # Si le score est suffisant, on met à jour la table Users
    if best_score >= 80:
        table_users.at[best_index, 'Mail'] = row['Mail1']

# Affichage du résultat
print("Table Users mise à jour :\n")
print(table_users[['Username', 'Login', 'Mail']])
```
